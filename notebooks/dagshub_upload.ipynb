# One-Time Setup: Upload Data to DagsHub via Colab
This notebook uses Colab's fast internet to download the dataset from Kaggle and upload it to your DagsHub repository using DVC. You only need to run this once!

```
!pip install kagglehub dvc dvc-s3 dagshub
import dagshub

# Authenticate with DagsHub
# This will prompt you for a token
dagshub.login()

# Initialize Git and DVC in Colab
!git init
!dvc init

# Download dataset from Kaggle
import os
import shutil
import kagglehub

print("Downloading dataset from Kaggle...")
os.environ["KAGGLEHUB_VERIFY_SSL"] = "0"
temp_path = kagglehub.dataset_download("andrewmvd/medical-mnist")

# Move to data/raw
os.makedirs("data/raw", exist_ok=True)
for item in os.listdir(temp_path):
    s = os.path.join(temp_path, item)
    d = os.path.join("data/raw", item)
    if os.path.isdir(s):
        shutil.copytree(s, d, dirs_exist_ok=True)
    else:
        shutil.copy2(s, d)
print("Data moved to data/raw")

# Track with DVC
!dvc add data/raw

# Configure DagsHub remote and push
# Replace Cludoy and DSAI-490-A1 if necessary
!dvc remote add origin https://dagshub.com/Cludoy/DSAI-490-A1.dvc
!dvc remote modify origin --local auth basic
!dvc remote modify origin --local user Cludoy

print("Pushing data to DagsHub DVC Storage...")
!dvc push -r origin

print("Done! Data is now tracked on DagsHub.")
```
